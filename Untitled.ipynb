```python
# Cell 2: Load the XES event log

from pathlib import Path
import pm4py

# Put BPI_Challenge_2019.xes in the same folder as this notebook,
# or replace this with the full path to the file on your computer.
file_path = Path("BPI_Challenge_2019.xes")

if not file_path.is_file():
    raise FileNotFoundError(
        f"Could not find {file_path.resolve()}. Place the XES file beside this notebook "
        "or update file_path to its actual location."
    )

log = pm4py.read_xes(str(file_path))

# Quick sanity check: how many rows (events) did we load
print(log.shape)
print(log.columns.tolist())
```

```python
# Cell 3: Understand the data - activities and basic counts

# Confirm case ID and activity columns (XES standard names)
case_id_col = 'case:concept:name'
activity_col = 'concept:name'

print("Number of cases:", log[case_id_col].nunique())
print("Number of events:", len(log))
print()
print("Unique activities in this log:")
print(log[activity_col].value_counts())
```

```python
# Pick one case ID and look at its full event sequence
sample_case_id = log['case:concept:name'].iloc[0]

sample_case = log[log['case:concept:name'] == sample_case_id][
    ['case:concept:name', 'concept:name', 'time:timestamp', 'org:resource']
].sort_values('time:timestamp')

print(sample_case)
```

```python
# Cell 5: Take a reproducible sample of cases

import pandas as pd

# Get list of all unique case IDs
all_case_ids = log['case:concept:name'].unique()
print("Total cases available:", len(all_case_ids))

# Randomly pick 3000 of them, but with a fixed seed so it's the same 3000 every time
sample_size = 3000
rng = pd.Series(all_case_ids).sample(n=sample_size, random_state=42)
sampled_case_ids = rng.tolist()

# Filter the full log down to only events belonging to these sampled cases
log_sample = log[log['case:concept:name'].isin(sampled_case_ids)].copy()

print("Sampled cases:", log_sample['case:concept:name'].nunique())
print("Sampled events:", len(log_sample))
```

```python
# Cell 6: Basic data quality checks on the sample

# 1. Check for missing values in the key columns we care about most
key_cols = ['case:concept:name', 'concept:name', 'time:timestamp']
print("Missing values in key columns:")
print(log_sample[key_cols].isnull().sum())
print()

# 2. Check for any duplicate rows (exact same case, activity, timestamp repeated)
duplicates = log_sample.duplicated(subset=['case:concept:name', 'concept:name', 'time:timestamp']).sum()
print("Exact duplicate rows:", duplicates)
print()

# 3. Check timestamp range, do dates look sane (not in the future, not some weird default date)
print("Earliest timestamp:", log_sample['time:timestamp'].min())
print("Latest timestamp:", log_sample['time:timestamp'].max())
```

```python
# Cell 7 (rearranged): Inspect duplicate-flagged rows with clear column order
import pandas as pd
pd.set_option('display.width', 1000)
pd.set_option('display.max_columns', None)

dup_mask = log_sample.duplicated(subset=['case:concept:name', 'concept:name', 'time:timestamp'], keep=False)
dup_rows = log_sample[dup_mask].sort_values(['case:concept:name', 'time:timestamp']).reset_index()

dup_rows = dup_rows.rename(columns={
    'index': 'Row',
    'case:Purchasing Document': 'PO',
    'case:concept:name': 'Case',
    'concept:name': 'Activity',
    'time:timestamp': 'Timestamp'
})

display(dup_rows[['Row', 'PO', 'Case', 'Activity', 'Timestamp']].head(20))
```

```python
import pandas as pd
pd.set_option('display.width', 1000)
pd.set_option('display.max_columns', None)

before = len(log_sample)

# display(log_sample)

log_sample = log_sample.drop_duplicates(subset=['case:concept:name', 'concept:name', 'time:timestamp', 'case:Item'])
after = len(log_sample)

print("Rows before:", before)

print("Rows after:", after)
print("Dropped:", before - after)

# display(log_sample)
```

```python
# Cell 9: Discover and visualize the process map (DFG)

# pm4py needs the log formatted a specific way for discovery algorithms
# Rename the columns to pm4py's expected internal names, then format it
log_sample_formatted = pm4py.format_dataframe(
    log_sample,
    case_id='case:concept:name',
    activity_key='concept:name',
    timestamp_key='time:timestamp'
)

# Discover the directly-follows graph
dfg, start_activities, end_activities = pm4py.discover_dfg(log_sample_formatted)

# Visualize it
pm4py.view_dfg(dfg, start_activities, end_activities)
```

```python
# Cell 9b: Manually filter to the most frequent activities

# Count how often each activity appears, keep only the top N most common
top_n = 12
top_activities = log_sample_formatted['concept:name'].value_counts().head(top_n).index.tolist()
print("Keeping these activities:", top_activities)

# Filter the log down to only events using these top activities
log_filtered = log_sample_formatted[log_sample_formatted['concept:name'].isin(top_activities)]

# Discover and view the process map on this filtered log
dfg2, start2, end2 = pm4py.discover_dfg(log_filtered)
pm4py.view_dfg(dfg2, start2, end2)
```

```python
# Cell 9c: Tighter filter - just the core P2P happy path activities

top_n = 7
top_activities = log_sample_formatted['concept:name'].value_counts().head(top_n).index.tolist()
print("Keeping these activities:", top_activities)

log_filtered = log_sample_formatted[log_sample_formatted['concept:name'].isin(top_activities)]

dfg3, start3, end3 = pm4py.discover_dfg(log_filtered)
pm4py.view_dfg(dfg3, start3, end3)

# Save the DFG as an actual PNG file instead of just viewing it
pm4py.save_vis_dfg(dfg3, start3, end3, 'process_map.png')
```
